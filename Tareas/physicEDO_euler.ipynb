# EDO A RESOLVER
$$
\frac{dN}{dt}=-\lambda*N
$$
Con condiciones iniciales N0=1000, Ecuacion tomada de la clase de Subatomica 2024-2 profesora Edna Carolina Pinilla Beltran, sistema fisico: **Decaimiento Radiactivo**

```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros físicos
lambda_ = 0.1    # constante de decaimiento
N0 = 1000         # número inicial de núcleos

# Parámetros numéricos
t0 = 0
tf = 100 
h = 0.5

# Número de pasos
n_steps = int((tf - t0) / h)

# Arrays
t = np.zeros(n_steps + 1)
N = np.zeros(n_steps + 1)

# Condición inicial
t[0] = t0
N[0] = N0

# Método de Euler
for i in range(n_steps):
    t[i+1] = t[i] + h
    N[i+1] = N[i] - h * lambda_ * N[i]

# Solución exacta 
N_exact = N0 * np.exp(-lambda_ * t)


# Gráfica
plt.figure()
plt.plot(t, N, label="Euler")
plt.plot(t, N_exact, linestyle="--", label="Exacta")
plt.xlabel("Tiempo (años)")
plt.ylabel("N(t)")
plt.title("Decaimiento Radiactivo (Euler vs Exacta)")
plt.legend()
plt.grid()

plt.show()
```
